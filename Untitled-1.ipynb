```python
import json
import numpy as np
import matplotlib.pyplot as plt
import os
import torch
from scipy import signal # This is for signal processing

# --- Use your exact paths ---
# This is the root folder containing all the modulation types
DATA_ROOT = '/Users/MAC/Downloads/5 GHz Bandwidth'

# For this example, we'll just pick one of the sub-folders to inspect
SAMPLE_DIR = os.path.join(DATA_ROOT, '64QAM_2.5GSyms_600mV')

print("Paths defined successfully!")
print(f"Data will be loaded from: {SAMPLE_DIR}")
```

```python
# We'll look at 'frame_1' as our test subject
meta_file = 'frame_3.sigmf-meta'
data_file = 'frame_3.sigmf-data'

# Construct the full paths to the files
meta_path = os.path.join(SAMPLE_DIR, meta_file)
data_path = os.path.join(SAMPLE_DIR, data_file)

# --- 1. Get the label from the folder path ---
# os.path.basename gets the last part of the path ('4PSK_2.5GSyms_75mV')
folder_name = os.path.basename(SAMPLE_DIR)
# We split by '_' and take the first part ('4PSK')
modulation_type = folder_name.split('_')[0]

print("--- Data Label ---")
print(f"Modulation Type (from path): {modulation_type}")

# --- 2. Load the metadata ONLY to confirm the data type ---
with open(meta_path, 'r') as f:
    metadata = json.load(f)
print(f"Data Format Hint (from global): {metadata['global']['core:datatype']}") # 'rf32_le'

# --- 3. Load the raw signal data ---
waveform_1d = np.fromfile(data_path, dtype=np.float32)

print("\n--- Data Shape ---")
print(f"Shape of the loaded 1D waveform: {waveform_1d.shape}")
```

```python
# --- Read just the first 100 numbers from the binary data file ---

# The np.fromfile function has a 'count' argument that lets us specify
# exactly how many items to read.
count_to_read = 100

# We use the same function as before, but now with the count argument.
# This is much more efficient than loading all 93,440 numbers if we only need a few.
first_100_values = np.fromfile(data_path, dtype=np.float32, count=count_to_read)


# --- Display the results ---
print(f"--- Successfully read the first {count_to_read} numbers ---")
print(first_100_values)

# Let's also confirm the shape of the resulting array
print(f"\nShape of the loaded array: {first_100_values.shape}")
```

```python
# --- STFT Parameters ---
nperseg = 256
noverlap = 128

# --- Perform the STFT ---
f, t, Zxx = signal.stft(waveform_1d, fs=1, nperseg=nperseg, noverlap=noverlap)

# --- Process the STFT result for the CNN ---
spectrogram_2d = np.abs(Zxx)
spectrogram_2d_with_channel = np.expand_dims(spectrogram_2d, axis=0)

print(f"Shape of the final 2D spectrogram: {spectrogram_2d_with_channel.shape}")
```

```python
# For better visualization, we'll plot the spectrogram on a log scale
log_spectrogram = 10 * np.log10(spectrogram_2d + 1e-9)

fig, axs = plt.subplots(2, 1, figsize=(12, 10))

# Plot the original raw waveform
axs[0].plot(waveform_1d[:1000])
axs[0].set_title("Original 1D Waveform (Zoomed In)")
axs[0].grid(True)

# Plot our new 2D spectrogram
im = axs[1].pcolormesh(t, f, log_spectrogram, shading='gouraud')
axs[1].set_title(f"Generated 2D Spectrogram")
axs[1].set_ylabel('Frequency Bins')
axs[1].set_xlabel('Time Bins')
fig.colorbar(im, ax=axs[1], label='Power [dB]')

plt.tight_layout()
plt.show()
```
